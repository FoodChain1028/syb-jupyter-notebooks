# SYB Network - Game Mode 🎮

A simple, terminal-like interface that runs like a game!

## How it works:
- ⏱️ Every 10 seconds, a random user makes a transaction (automatic)
- ⚡ When queue reaches 5 transactions, a batch is automatically forged
- 🎮 You can add your own transactions anytime while it runs
- 📺 All updates print to the output in real-time

## No widgets, just prints - works everywhere! ✨

## 1. Setup

Import and create the network.

````python
# Enable nested asyncio for Jupyter
import nest_asyncio
nest_asyncio.apply()

from syb_game import create_random_network, start_game
import asyncio

print("✅ Imports successful!")
````

````python
# Create network
contract, users = create_random_network(
    num_users=8,
    random_name='erdos_renyi',
    balance_range=(0.01, 0.1)
)

print(f"\n✅ Network ready with {len(users)} users!")
````

## 2. Start the Game

This will start the automatic simulation.

**What you'll see:**
- Every 10 seconds: New transaction
- When queue reaches 5: Batch forges automatically
- Status updates after each event

````python
# Start the game (10 second intervals)
game, task = await start_game(contract, users, interval=5)

# The game is now running in the background!
# Watch the output above - transactions will appear every 10 seconds
````

## 3. Manual Actions

While the game runs, you can interact anytime!

### Available Commands:

- `game.my_balance()` - Check your balance and score
- `game.list_users()` - See all users
- `game.show_queue()` - View transaction queue
- `game.deposit(1.0)` - Deposit 1.0 ETH
- `game.withdraw(0.5)` - Withdraw 0.5 ETH
- `game.vouch("User 1")` - Vouch for User 1
- `game.unvouch("User 1")` - Unvouch User 1
- `game.create_user()` - Create a new user

Try them in the cells below!

````python
# Check your balance
game.my_balance()
````

````python
# List all users
game.list_users()
````

````python
# Show current queue
game.show_queue()
````

````python
# Add your own transaction - deposit 1 ETH
game.deposit(1.0)
````

````python
# Vouch for another user
game.vouch("User 1")
````

````python
# Create a new user
game.create_user()
````

## 5. Stop the Game

When you're done watching, stop the game:

````python
# Stop the game
task.cancel()
game.stop()

print("\n🛑 Game stopped!")
````

## 6. Quick Start (All in One Cell)

Want to restart? Run everything in one cell:

````python
# Quick restart
import nest_asyncio
nest_asyncio.apply()

from syb_game import create_random_network, start_game
import asyncio

# Create and start
contract, users = create_random_network(num_users=8, balance_range=(0.01, 0.1))
game, task = await start_game(contract, users, interval=10)

print("\n🎮 Game running! Use game.my_balance(), game.deposit(1.0), etc.")
print("   Watch above for automatic transactions every 10 seconds!")
````
