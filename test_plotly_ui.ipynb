# SYB Network UI - Plotly Version (Binder-Friendly)

This notebook demonstrates the simplified UI implementation using Plotly for interactive graphs.

**Features:**
- ✅ Interactive Plotly network graphs (works on Binder)
- ✅ Manual refresh controls (no threading issues)
- ✅ Simulation buttons for testing
- ✅ Clean, simple interface

**How to use:**
1. Run all cells to initialize the network
2. Use the UI buttons to interact with the network
3. Click "Refresh All" to update displays after actions
4. Use "Simulate Transaction" to generate random activity

## 1. Setup and Imports

```python
# Import the Plotly UI module
from syb_ui_plotly import create_random_network, SYBUserInterfacePlotly

print("✅ Imports successful!")
```

## 2. Create Random Network

This creates a random Erdos-Renyi network with 8 users and initial connections.

```python
# Create network with 8 users
contract, users = create_random_network(
    num_users=8, 
    random_name='erdos_renyi',
    balance_range=(0.01, 0.1)
)

print(f"\n✅ Network created with {len(users)} users!")
print(f"📊 Network has {contract.network.graph.number_of_nodes()} nodes and {contract.network.graph.number_of_edges()} edges")
```

## 3. Launch Interactive UI

This will display the full interactive interface with:
- Transaction queue viewer
- Network status display
- User action buttons (deposit, withdraw, vouch, unvouch)
- Interactive Plotly graph
- Simulation controls

```python
# Create and display the UI
ui = SYBUserInterfacePlotly(contract, users)
ui.display()
```

## 4. Manual Testing (Optional)

You can also interact with the contract programmatically:

```python
# Example: Check current queue size
print(f"Current queue size: {len(contract.unprocessed_txns)}")
print(f"Batch size: {contract.batch_size}")
print(f"Last forged batch: #{contract.last_forged_batch}")
```

```python
# Example: View all user balances
print("User Balances:")
print("=" * 50)
for addr, user_data in users.items():
    balance = user_data['interface'].get_my_balance() / 10**18
    score = user_data['interface'].get_my_score()
    print(f"{user_data['name']:10} | Balance: {balance:8.4f} ETH | Score: {score:.4f}")
```

```python
# Example: Manually add some transactions to test batch forging
user_addrs = list(users.keys())

# Add 5 transactions to fill a batch
for i in range(5):
    user = users[user_addrs[i % len(user_addrs)]]
    user['interface'].deposit(int(0.1 * 10**18))
    
print(f"✅ Added 5 transactions. Queue size: {len(contract.unprocessed_txns)}")
print("💡 Now click 'Forge Batch' in the UI above, then 'Refresh All'")
```

## Tips for Using on Binder

1. **Manual Refresh**: Unlike the threaded version, this UI requires you to click "Refresh All" to see updates
2. **Simulation**: Use "Simulate Transaction" to generate random activity without manual input
3. **Batch Processing**: When queue reaches batch size (5), click "Forge Batch" to process transactions
4. **Interactive Graph**: Hover over nodes to see details, zoom and pan as needed
